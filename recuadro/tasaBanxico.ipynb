```python
import requests
import pandas as pd
from datetime import date

TOKEN = "629249a37423850587476dbd81dca910635fcf1df085430f80c788a09d34c9e8"

# Serie: Tasa objetivo
serie = "SF61745"

inicio = "1998-01-01"
fin = date.today().strftime("%Y-%m-%d")  # fecha de hoy en formato yyyy-mm-dd

url = f"https://www.banxico.org.mx/SieAPIRest/service/v1/series/{serie}/datos/{inicio}/{fin}"

headers = {"Bmx-Token": TOKEN}

response = requests.get(url, headers=headers)
print("Status:", response.status_code)

data = response.json()

# Extraer datos
datos = data["bmx"]["series"][0]["datos"]

df = pd.DataFrame(datos)
df.columns = ["fecha", "tasa_objetivo"]

# OJO: Banxico suele devolver fecha como dd/MM/yyyy dentro del JSON
df["fecha"] = pd.to_datetime(df["fecha"], format="%d/%m/%Y")
df["tasa_objetivo"] = pd.to_numeric(df["tasa_objetivo"], errors="coerce")

df = df.sort_values("fecha").reset_index(drop=True)

df.head()
```

```python
import requests
import pandas as pd
from datetime import date

TOKEN = "629249a37423850587476dbd81dca910635fcf1df085430f80c788a09d34c9e8"


serie = "SF60633"  # CETES 28 días

inicio = "1998-01-01"
fin = date.today().strftime("%Y-%m-%d")

url = f"https://www.banxico.org.mx/SieAPIRest/service/v1/series/{serie}/datos/{inicio}/{fin}"

headers = {"Bmx-Token": TOKEN}

response = requests.get(url, headers=headers)
print("Status:", response.status_code)

data = response.json()
datos = data["bmx"]["series"][0]["datos"]

df_cetes = pd.DataFrame(datos)
df_cetes.columns = ["fecha", "cetes_28"]

df_cetes["fecha"] = pd.to_datetime(df_cetes["fecha"], format="%d/%m/%Y")
df_cetes["cetes_28"] = pd.to_numeric(df_cetes["cetes_28"], errors="coerce")

df_cetes = df_cetes.sort_values("fecha").reset_index(drop=True)

df_cetes.head()
```
